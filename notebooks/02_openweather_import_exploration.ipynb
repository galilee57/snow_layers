```python
import pandas as pd

path = '/content/drive/MyDrive/ColabFiles/Snow_Meribel_1.csv'
df_snow = pd.read_csv(path, sep=';')
```

```python
df_snow.columns
```

```python
# Conversion de la colonne datetime en type datetime
df_snow['datetime'] = pd.to_datetime(df_snow['datetime'])
# On va utiliser la colonne datetime (desormais en type datetime) pour indexer df_snow
df_snow_time = df_snow.set_index('datetime')
```

```python
# On peut ensuite la manipuler avec des méthodes
df_snow['weekday'] = df_snow['datetime'].dt.day_name()
df_snow['weekday'][:2]
```

```python
# Avec loc, on peut slicer le dataframe.
df_snow_time.loc['2022-01-12':'2022-01-13','tempmax']
```

```python
# Avec la méthode resample, on peut calculer des statistiques sur des périodes (D, M, Y, ...).
df_snow_time['snowdepth'].resample('M').mean()
```

```python
from plotly.subplots import make_subplots
import plotly.graph_objects as go
fig = make_subplots(rows=3, cols=1, shared_xaxes=True)

snow_graph = go.Scatter(x = df_snow_time.index, y=df_snow_time['snow'], name='snow')
precip_graph = go.Scatter(x = df_snow_time.index, y=df_snow_time['precip'], name='precip')
snowdepth_graph = go.Scatter(x = df_snow_time.index, y=df_snow_time['snowdepth'], name='snowdepth')

fig.add_trace(snow_graph, row=1, col=1)
fig.add_trace(precip_graph, row=1, col=1)
fig.add_trace(snowdepth_graph, row=2, col=1)
```
